# 2D Massenschwinger

Für das dargestellte System soll die Bewegungsgleichung aufgestellt werden und die Eigenfrequenzen und Eigenvektoren berechnet werden.

![2DMass](spring_2.png)

```python
import sympy as sp
from sympy import symbols,Matrix,Function,Eq

m_1,m_2,k_1,k_2,k_3,t,m,k = symbols("m_1 m_2 k_1 k_2 k_3 t,m,k",positive=True)

soldict = {
    m_1:3*m,
    m_2:m,
    k_1:k,
    k_2:2*k,
    k_3:3*k
}

# soldict = {
#     m_1:1,
#     m_2:2,
#     k_1:2000,
#     k_2:3000,
#     k_3:1000
# }

u_1 = Function("u_1")(t)
u_2 = Function("u_2")(t)

u1_dd = u_1.diff(t,t)
u2_dd = u_2.diff(t,t)

ode1= Eq(-m_1 * u1_dd - (k_2+k_1) * u_1 + k_2 * u_2,rhs=0)
ode2 = Eq(-m_2 * u2_dd - (k_2+k_3)*u_2 + k_2 * u_1,rhs=0)

display(ode1)
display(ode2)

```

### In Matrixform lautet die Gleichung

```python
M = Matrix(
    [[m_1,0],[0,m_2]]
)
ddu = Matrix(
    [[u1_dd],[u2_dd]]
)
K = Matrix(
    [[k_1+k_2,-k_2],[-k_2,k_2+k_3]]
)
u = Matrix(
    [[u_1],[u_2]]
)
display(M.subs(soldict))
display(K.subs(soldict))
```

### Bestimmung der Eigenwerte und Eigenvektoren

```python
from sympy import simplify,latex
D = M.inv() * K

# Calculate eigenvalues, which are the squares of the angular frequencies
B = D.subs(soldict)
eigenvals = simplify(B.eigenvals())

eigenvecs = B.eigenvects()

# Print the eigenvalues
print("Eigenvalues:")
for val in eigenvals:
    display(val)
    print(latex(val))
    
# Print the eigenvecs
print("Eigenvecs:")
for eigenval, multiplicity, eigenvec in eigenvecs:
    display(eigenvec[0])
    print(latex(eigenvec[0]))
```

### Bestimmung der effektiven Massen

```python
phi_1 = eigenvecs[0][2][0]
phi_2 = eigenvecs[1][2][0]
phi = Matrix([phi_1.T,phi_2.T])
phi
```

```python
M = M.subs(soldict)
gamma = phi*M*Matrix([[1],[1]])
gamma
```

```python
print(latex(phi_2.subs(soldict)))
```

```python
gamma_1 = simplify(phi_1.T*M*Matrix([[1],[1]]))
display("gamma_1:",gamma_1)
meff_1 = simplify(gamma_1**2/(phi_1.T*M*phi_1)[0])[0,0]
simplify(meff_1)
rel_meff_1=(simplify((meff_1/(m_1+m_2)).subs(soldict)))
print(f'gamma_1: {latex(gamma_1)}')
print(f'meff_1: {latex(meff_1)}')
print(f'relative mass: {rel_meff_1}')
display(float(rel_meff_1))
```

```python
gamma_2 = simplify(phi_2.T*M*Matrix([[1],[1]]))
display("gamma_2:",gamma_2)
meff_2 = simplify(gamma_2**2/(phi_2.T*M*phi_2)[0])[0,0]
simplify(meff_2)
rel_meff_2=(simplify((meff_2/(m_1+m_2)).subs(soldict)))

print(f'gamma_2: {latex(gamma_2)}')
print(f'meff_2: {latex(meff_2)}')
print(f'relative mass: {rel_meff_2}')
display(float(rel_meff_2))
```
